```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns; sns.set() # sns.set() ==> グラフの見た目をseabornに合わせる
import japanize_matplotlib
import datetime as dt
%matplotlib inline
# CSVファイルを読み込む
df = pd.read_csv('../utf8.csv')

# 面積データを取得
areas = df['area_building']

# ヒストグラムを作成
plt.hist(areas, bins=50, edgecolor='black')
plt.xlabel('建物面積 (m²)')
plt.ylabel('建物数')
plt.title('建物面積の分布')
plt.grid(True)
plt.show()
```

```python
# 箱ひげ図を作成
plt.figure(figsize=(10, 6))
sns.boxplot(x=areas)
plt.xlabel('建物面積 (m²)')
plt.title('建物面積の箱ひげ図')
plt.grid(True)
plt.show()

# 累積分布を作成
plt.figure(figsize=(10, 6))
sns.ecdfplot(data=areas)
plt.xlabel('建物面積 (m²)')
plt.ylabel('累積確率')
plt.title('建物面積の累積分布')
plt.grid(True)
plt.show()

# 基本統計量を表示
print('基本統計量:')
print(f'平均値: {areas.mean():.2f} m²')
print(f'中央値: {areas.median():.2f} m²')
print(f'標準偏差: {areas.std():.2f} m²')
print(f'最小値: {areas.min():.2f} m²')
print(f'最大値: {areas.max():.2f} m²')
print(f'第1四分位数: {areas.quantile(0.25):.2f} m²')
print(f'第3四分位数: {areas.quantile(0.75):.2f} m²')
```

```python
import numpy as np

# 面積データを取得（500m²以下にフィルタリング）
areas = df[df['area_building'] <= 500]['area_building']

# 3つのグラフを横に並べて表示
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))

# ヒストグラム
ax1.hist(areas, bins=50, edgecolor='black')
ax1.set_xlabel('建物面積 (m²)')
ax1.set_ylabel('建物数')
ax1.set_title('建物面積の分布 (500m²以下)')
ax1.grid(True)

# 箱ひげ図
sns.boxplot(y=areas, ax=ax2)
ax2.set_ylabel('建物面積 (m²)')
ax2.set_title('建物面積の箱ひげ図 (500m²以下)')

# 累積分布
areas_sorted = np.sort(areas)
p = np.arange(1, len(areas) + 1) / len(areas)
ax3.plot(areas_sorted, p)
ax3.set_xlabel('建物面積 (m²)')
ax3.set_ylabel('累積確率')
ax3.set_title('建物面積の累積分布 (500m²以下)')
ax3.grid(True)

plt.tight_layout()
plt.show()

# 基本統計量を表示
print("\n基本統計量:")
print(areas.describe())
```
