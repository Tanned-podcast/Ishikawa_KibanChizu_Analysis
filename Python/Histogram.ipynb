```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns; sns.set() # sns.set() ==> グラフの見た目をseabornに合わせる
import japanize_matplotlib
import datetime as dt
%matplotlib inline
# フォントの設定
plt.rcParams['font.family'] = 'IPAexGothic'
# グローバルにフォントサイズを設定
plt.rcParams.update({'font.size': 16})

# CSVファイルを読み込む
df = pd.read_csv('../building_area_csv/building_area_summary.csv')

# 面積データを取得
areas = df["area[m2]"]

# ヒストグラムを作成
plt.hist(areas, bins=50, edgecolor='black')
plt.xlabel('建物面積 (m²)')
plt.ylabel('建物数')
plt.title('建物面積の分布')
plt.grid(True)
plt.show()

# 箱ひげ図を作成
plt.figure(figsize=(10, 6))
sns.boxplot(x=areas)
plt.xlabel('建物面積 (m²)')
plt.title('建物面積の箱ひげ図')
plt.grid(True)
plt.show()

# 累積分布を作成
plt.figure(figsize=(10, 6))
sns.ecdfplot(data=areas)
plt.xlabel('建物面積 (m²)')
plt.ylabel('累積確率')
plt.title('建物面積の累積分布')
plt.grid(True)
plt.show()

# 基本統計量を表示
print('基本統計量:')
print(f'平均値: {areas.mean():.2f} m²')
print(f'中央値: {areas.median():.2f} m²')
print(f'標準偏差: {areas.std():.2f} m²')
print(f'最小値: {areas.min():.2f} m²')
print(f'最大値: {areas.max():.2f} m²')
print(f'第1四分位数: {areas.quantile(0.25):.2f} m²')
print(f'第3四分位数: {areas.quantile(0.75):.2f} m²')
print(f'95%: {areas.quantile(0.95):.2f} m²')
```

```python
#次はDRMのデータから道路幅の分布を出す
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns; sns.set() # sns.set() ==> グラフの見た目をseabornに合わせる
import japanize_matplotlib
import datetime as dt
%matplotlib inline
# フォントの設定
plt.rcParams['font.family'] = 'IPAexGothic'
# グローバルにフォントサイズを設定
fontsize = 16

# CSVファイルを読み込む
df = pd.read_csv('../road_csv/DRM_road_data.csv')

# 面積データを取得
roads = df["道路幅員区分コード"]

# 道路幅員区分コードに対応するラベルを定義（例）
width_labels = {
    0: 'No Data',
    1: 'd>=13[m]',
    2: '5.5=<d<13[m]',
    3: '3.0=<d<5.5[m]',
    4: 'd<3.0[m]'
}

# 区分コードでカウント
code_counts = roads.value_counts().sort_index()  # ここで数値順に並べる

# 対応するラベルに変換（順番保持）
labels = [width_labels.get(code, str(code)) for code in code_counts.index]
counts = code_counts.values

# グラフ描画
plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=counts, palette="Blues_d", edgecolor="black")
plt.xlabel('道路幅員区分', fontsize=fontsize)
plt.ylabel('リンク数', fontsize=fontsize)
plt.title('DRM道路データの道路幅員分布', fontsize=fontsize)
plt.xticks(rotation=30, ha='right', fontsize=fontsize)
plt.yticks(fontsize=fontsize)
plt.grid(True, axis='y')
plt.tight_layout()
plt.show()

# 基本統計量を表示
print('各区分リンク数:')
print(f'd>=13[m]: {counts[0]}')
print(f'5.5=<d<13[m]: {counts[1]}')
print(f'3.0=<d<5.5[m]: {counts[2]}')
print(f'd<3.0[m]: {counts[3]}')
```

```python
# フォント設定
plt.rcParams['font.family'] = 'IPAexGothic'
fontsize = 24  # グローバルにフォントサイズを設定

# CSVファイルを読み込む
df = pd.read_csv('../road_csv/DRM_road_data.csv')

# 幅員区分コードとリンク長データを取得
width_code = df["道路幅員区分コード"]
link_length = df["リンク長（計算値）"]  # 単位が m か km かはデータで確認

# 幅員区分コードとリンク長を使ってグループごとの総延長を計算
grouped = ((df.groupby("道路幅員区分コード")["リンク長（計算値）"].sum())/1000).sort_index()

# ラベルとデータをリストとして準備
labels = [width_labels.get(code, str(code)) for code in grouped.index]
lengths = grouped.values  # 総延長の配列

# グラフ描画
plt.figure(figsize=(10, 6))
sns.barplot(x=labels, y=lengths, palette="Greens_d", edgecolor="black")
plt.xlabel('道路幅員区分', fontsize=fontsize)
plt.ylabel('リンク総延長[km]', fontsize=fontsize)
plt.title('道路幅員区分ごとのリンク総延長', fontsize=fontsize)
plt.xticks(rotation=30, ha='right', fontsize=fontsize)
plt.yticks(fontsize=fontsize)
plt.grid(True, axis='y')
plt.tight_layout()
plt.show()

# 基本統計量を表示
print('各区分リンク総延長:')
print(f'd>=13[m]: {lengths[0]}')
print(f'5.5=<d<13[m]: {lengths[1]}')
print(f'3.0=<d<5.5[m]: {lengths[2]}')
print(f'd<3.0[m]: {lengths[3]}')
```
